```python
from torch.utils.data import Dataset
from PIL import Image
import torchvision.transforms as transforms

class CustomDataset(Dataset):
    def __init__(self, df):
        self.df = df
        self.transform = transforms.Compose([
            transforms.Resize((224,224)),
            transforms.ToTensor()
        ])

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        img_path = self.df.iloc[idx]["image_path"]
        labels = self.df.iloc[idx]["Cluster"]
        # print(img_path)  

        img = Image.open(img_path).convert("RGB")
        img = self.transform(img)

        return img, labels

from torch.utils.data import DataLoader

train_dataset = CustomDataset(train_df)
#train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)

val_dataset = CustomDataset(val_df)
#val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)

test_dataset = CustomDataset(test_df)
#test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

```python
# build model
from torchvision.models import mobilenet_v3_small, MobileNet_V3_Small_Weights
from efficientnet_lite import build_efficientnet_lite
from torchvision.models import squeezenet1_1, SqueezeNet1_1_Weights
from torchvision.models import shufflenet_v2_x0_5, ShuffleNet_V2_X0_5_Weights
model_names = ['mobilenet','squeezenet'] #,'shufflenet','efficientnet'] --> notebook satunya lg

def build_model(model_name, device, dropout):
    # === mobilenet ===
    if model_name == 'mobilenet': 
        #load model
        model = mobilenet_v3_small(weights = MobileNet_V3_Small_Weights.IMAGENET1K_V1)

        # modify the final layer (classifier) to make it has an output of 5 classes
        in_features_mobilenet = model.classifier[-1].in_features # ambil jumlah input feature dari layer terakhir

        model.classifier[-1] = nn.Sequential( #replace final layer
            nn.Dropout(p=dropout), #replace dropout
            nn.Linear(in_features_mobilenet, 5) # ubah jml kelas pada classifier menjadi 5
        )

    # === efficientnet ===
    elif model_name == 'efficientnet':
        # build model 
        model = build_efficientnet_lite('efficientnet_lite0', 1000) # 1000 features to load the weight. 1000 karena data yang digunakan untuk melatih model pretrained (which is yang membentuk weight model saat ini), yakni ImageNet, memiliki 1000 kelas

        # load weight
        model.load_state_dict(torch.load('efficientnet_lite0.pth', map_location = device))

        # modify classifier
        in_features_efficientnet = model.fc.in_features 
        model.fc = nn.Sequential(
            nn.Dropout(p=dropout),
            nn.Linear(in_features_efficientnet, 5)
        )
    
    # === squeezenet ===
    elif model_name == 'squeezenet':
        # load model 
        model = squeezenet1_1(weights = SqueezeNet1_1_Weights.IMAGENET1K_V1)

        # modify the final layer (classifier) to make it has an output of 5 classes
        in_channels_squeezenet_v1_1 = model.classifier[1].in_channels # ambil jumlah input channel (kalo conv2d pake channels, bukan features) dari layer terakhir

        model.classifier[1] = nn.Conv2d( # pake 1, bukan -1. but why?
            in_channels_squeezenet_v1_1, 
            5,
            kernel_size=(1,1) # kenapa pake ukuran kernel 1,1? 
        )

        # tambahkan layer dropout di depan classifier. layer dropout harus ditaruh di sini karena kalo dropout dulu baru nn.Conv2d, nanti overwritten shg dropout yang kita tambahin hilang begitu saja 
        model.classifier = nn.Sequential(
            nn.Dropout(p=dropout), # layer baru
            *model.classifier # iterate over semua layer lama
        )
        '''
        output:
        Dropout (baru)
        Dropout (lama)
        Conv2d
        ReLU
        AdaptiveAvgPool2d
        '''        
    
    # === shufflenet ===
    elif model_name == 'shufflenet':
        #load model
        model = shufflenet_v2_x0_5(weights = ShuffleNet_V2_X0_5_Weights.IMAGENET1K_V1)

        # modify the final layer (classifier) to make it has an output of 5 classes
        in_features_shufflenetv2x05 = model.fc.in_features #  Untuk shufflenet, pakenya bukan "classifier[-1]", tetapi "fc" (fully connected. basically classifier)

        model.fc = nn.Sequential( #replace final layer
            nn.Dropout(p=dropout), #replace dropout
            nn.Linear(in_features_shufflenetv2x05, 5) # ubah jml kelas pada classifier menjadi 5
        )

    return model
```

```python
import copy
import time

# training and validation function

def train_and_evaluate (model_name, model, train_loader, val_loader, optimizer, criterion, device, l1_lambda, l2_lambda):
    model.to(device)
    
    # create empty list to store useful value (it's helpful for tracking model progress)
    epoch_count, train_loss_values, val_loss_values, f1_train_values, f1_val_values = [], [], [], [], []
    
    # inisiasi variabel.#since kita mau maksimalin f1, maka kita gak cari loss paling minimum, merely nyari nilai loss ketika f1 maksimum (ketahui bahwa f1 score maksimum tidak selalu berarti loss minimum)
    best_f1 = float("-inf")
    epoch_best_f1 = None
    val_loss_best_f1 = None
    train_loss_best_f1 = None
    best_model_state = None
    
    # calculate the number of weights in the model
    params_no_bias = [p for n, p in model.named_parameters() if 'bias' not in n] # model.named_parameters() me-return pasangan (name, parameters) yang dalam koode ini, dimisalkan sebagai n dan p
    
    # timer starts
    start_time = time.time()

    for epoch in range(100):
        # === TRAINING ===
        
        # reset nilai loss val dan train menjadi 0 setiap terjadi pengulangan epoch 
        val_loss = 0
        train_loss = 0
        
        # create empty list to store all_preds and all_labels per epoch when training model
        all_preds_training, all_labels_training = [], []

        
        # put model in training mode 
        model.train()
        
        # forward & backward propagation across batches 
        for imgs, labels in train_loader: # iterate over batches from DataLoader
            imgs, labels = imgs.to(device), labels.to(device)
            
            optimizer.zero_grad() # optimizer zero grad
            outputs = model(imgs) # forward pass
            if model_name == "squeezenet": # flatten untuk mengubah output model squeezenet dari [batch, jumlah kelas, 1, 1] menjadi [batch, jumlah kelas] yang bisa diterima oleh loss function crossentropy loss 
                outputs = outputs.view(outputs.size(0), -1)
            loss = criterion(outputs,labels) # calculate loss
            
            # apply elasticnet regularization
            '''
            for name, weights in model.named_parameters():
                if 'bias' not in name:
                    l1_weights_sum = torch.sum(torch.abs(weights))
                    l2_weight_sum = torch.sum(torch.square(weights))
                    l1_term = l1_term + l1_weights_sum
                    l2_term = l2_term + l2_weight_sum
            l1_term = l1_term/nweights
            l2_term = l2_term/nweights
            '''
            l1_term = sum(torch.sum(torch.abs(p)) for p in params_no_bias) / len(params_no_bias)
            l2_term = sum(torch.sum(torch.square(p)) for p in params_no_bias) / len(params_no_bias)

  
            loss = loss + l1_lambda * l1_term + l2_lambda * l2_term
            
            loss.backward() # backpropagation on the loss function
            optimizer.step() # gradient descent 
            train_loss += loss.item() # total loss dari keseluruhan batches. .item() mengubah tensor menjadi float
            
            preds = torch.argmax(outputs, dim=1).cpu().numpy() # Ambil kelas dengan probabilitas tertinggi dari output model, ubah jadi CPU (karena .numpy() hanya bisa dipakai di CPU), kemudian ubah jadi array NumPy 

            # accumulate predictions and labels into a single Python list across multiple batches
            all_preds_training.extend(preds)
            all_labels_training.extend(labels.cpu().numpy()) 
        avg_train_loss = train_loss/len(train_loader) # average loss across batches
                
        # calculate macro average train f1-score
        f1_train_epoch = f1_score(all_labels_training, all_preds_training, average='macro') # train f1 score hanya digunakan untuk plotting untuk dibandingkan dengan val, adapun f1 score yang dipake untuk memilih model adalah f1 score val
        
        
        # === VALIDATION ===
        # create empty list to store all_preds and all_labels per epoch when validating model
        all_preds_val, all_labels_val = [], []
        # put model in eval mode 
        model.eval()

        with torch.no_grad(): # turn on no_grad() for faster performance bc it disables gradient tracking functionality (gradient tracking is not needed for inference)
            for imgs, labels in val_loader:
                imgs, labels = imgs.to(device), labels.to(device)
                outputs = model(imgs) # forward pass
                if model_name == "squeezenet": # flatten untuk mengubah output model squeezenet dari [batch, jumlah kelas, 1, 1] menjadi [batch, jumlah kelas] yang bisa diterima oleh loss function crossentropy loss 
                    outputs = outputs.view(outputs.size(0), -1)
                    
                #calculate loss
                loss = criterion(outputs, labels) 
                val_loss += loss.item()

                preds = torch.argmax(outputs, dim=1).cpu().numpy() # Ambil kelas dengan probabilitas tertinggi dari output model, ubah jadi CPU (karena .numpy() hanya bisa dipakai di CPU), kemudian ubah jadi array NumPy 

                # accumulate predictions and labels into a single Python list across multiple batches
                all_preds_val.extend(preds)
                all_labels_val.extend(labels.cpu().numpy()) 
        avg_val_loss = val_loss/len(val_loader)
        
        # calculate macro average val f1-score
        f1_val_epoch = f1_score(all_labels_val, all_preds_val, average='macro')
        
        
        # choose best f1 score, train_loss_best_f1, dan val_loss_best_f1
        if f1_val_epoch > best_f1: 
            best_f1 = f1_val_epoch
            epoch_best_f1 = epoch
            train_loss_best_f1 = avg_train_loss #train_loss_when_f1_reached_its_max_value
            val_loss_best_f1 = avg_val_loss #val_loss_when_f1_reached_its_max_value
            best_model_state = copy.deepcopy(model.state_dict()) #copy.deepcopy(model.state_dict()) # menyimpan weight, bias, buffer (BatchNorm, dll.) setelah model dilatih dan dievaluasi. berbeda dengan best_config yang menyimpan konfigurasi hyperparameter (learning rate, optimizer, dll) model
        
        #simpan important values setiap epoch untuk plotting curve nanti
        epoch_count.append(epoch)
        train_loss_values.append(avg_train_loss)
        val_loss_values.append(avg_val_loss)
        f1_train_values.append(f1_train_epoch)
        f1_val_values.append(f1_val_epoch)
        # print out what's happening every 10 epochs 
        if (epoch == 0 or epoch == 100):
            print(f"Epoch: {epoch} | Categorical Cross Entropy Train Loss: {avg_train_loss:.4f} | Categorical Cross Entropy Val Loss: {avg_val_loss:.4f} | F1-Score = {f1_val_epoch:.4f}")
        
    # timer ends
    end_time = time.time()
    total_time_seconds = end_time - start_time
    
    return epoch_count, train_loss_values, val_loss_values, f1_train_values, f1_val_values, best_f1, epoch_best_f1, val_loss_best_f1, train_loss_best_f1, best_model_state, total_time_seconds
```

```python
# grid search

device = torch.device("cuda" if torch.cuda.is_available() else "cpu") 

model_names = ['mobilenet','squeezenet'] #, 'shufflenet', 'efficientnet']
dropouts = [0.3, 0.5]
batch_sizes = [8, 16, 32]
optimizers = ['adam', 'rmsprop']
learning_rates = [0.001, 0.0001]
# tambahan regularisasi untuk mencegah overfitting. teknik regularisasi yang digunakan adalah elastic net regularization (landasan: https://ijettjournal.org/Volume-73/Issue-10/IJETT-V73I10P108.pdf)
l1_lambda = [0.001, 0.01] # logarithmic scale
l2_lambda = [0.001, 0.01]  # logarithmic scale

best_result = {
    "mobilenet": {
        "f1": float("-inf"), # menyimpan f1 score terbaik
        "config": None, # menyimpan konfigurasi hyperparameter terbaik
        "model_state": None, # menyimpan model state (weight, bias, dsb.) terbaik
        "epoch_count": None,
        "train_loss_values": None, # menyimpan history of train loss value dari model terbaik 
        "val_loss_values": None, # menyimpan history of val loss value dari model terbaik 
        "f1_train_values": None, # menyimpan history of train f1 value dari model terbaik 
        "f1_val_values": None, # menyimpan history of val f1 value dari model terbaik 
        "train_val_duration": None, # menyimpan durasi training dan validasi
    },
    "squeezenet": {
        "f1": float("-inf"),
        "config": None,
        "model_state": None,
        "epoch_count": None,
        "train_loss_values": None, # menyimpan history of train loss value dari model terbaik 
        "val_loss_values": None, # menyimpan history of val loss value dari model terbaik 
        "f1_train_values": None, # menyimpan history of train f1 value dari model terbaik 
        "f1_val_values": None, # menyimpan history of val f1 value dari model terbaik 
        "train_val_duration": None, # menyimpan durasi training dan validasi
    },
    "shufflenet": {
        "f1": float("-inf"),
        "config": None,
        "model_state": None,
        "epoch_count": None,
        "train_loss_values": None, # menyimpan history of train loss value dari model terbaik 
        "val_loss_values": None, # menyimpan history of val loss value dari model terbaik 
        "f1_train_values": None, # menyimpan history of train f1 value dari model terbaik 
        "f1_val_values": None, # menyimpan history of val f1 value dari model terbaik 
        "train_val_duration": None, # menyimpan durasi training dan validasi
    },
    "efficientnet": {
        "f1": float("-inf"),
        "config": None,
        "model_state": None,
        "epoch_count": None,
        "train_loss_values": None, # menyimpan history of train loss value dari model terbaik 
        "val_loss_values": None, # menyimpan history of val loss value dari model terbaik 
        "f1_train_values": None, # menyimpan history of train f1 value dari model terbaik 
        "f1_val_values": None, # menyimpan history of val f1 value dari model terbaik 
        "train_val_duration": None, # menyimpan durasi training dan validasi
    }
}

for model_name in model_names:
    # pastikan folder-nya ada
    base_dir = model_dirs[model_name]
    os.makedirs(base_dir, exist_ok=True)

    models_dir = os.path.join(base_dir, "models")
    plots_dir = os.path.join(base_dir, "plots")
    logs_dir = os.path.join(base_dir, "logs")

    os.makedirs(models_dir, exist_ok=True)
    os.makedirs(plots_dir, exist_ok=True)
    os.makedirs(logs_dir, exist_ok=True)
    log_file = os.path.join(logs_dir, f"results_{model_name}.json")    
    
    # load dulu kalau ada
    if os.path.exists(log_file):
        with open(log_file, "r") as f:
            results = json.load(f)

        if results:
            best_key = max(results, key=lambda k: results[k]["best_f1"])

            best_result[model_name]["f1"] = results[best_key]["best_f1"]
            best_result[model_name]["config"] = results[best_key]["config"]

            best_model_path_json = results[best_key]["model_path"]

            best_result[model_name]["model_state"] = torch.load(
                best_model_path_json,
                map_location=device
            )

    else:
        results = {}
        
    for dropout in dropouts:
        for batch_size, opt_name, lr, l1, l2 in product(batch_sizes, optimizers, learning_rates, l1_lambda, l2_lambda):            
            config_dict = {
            "bs": batch_size,
            "do": dropout,
            "opt": opt_name,
            "lr": lr,
            "l1": l1,
            "l2": l2,
            "model": model_name
            }
            config_key = "_".join([
            f"{k}={config_dict[k]:.6g}" if isinstance(config_dict[k], float) else f"{k}={config_dict[k]}"
            for k in sorted(config_dict)
            ])        

            # 🚫 SKIP kalau sudah pernah dijalankan
            if config_key in results: # jangan pake config_id, langsung sebut nama hyperparameter beserta nilainya dan nama model
                print(f"⏭️ Skip: {config_key}")
                continue
                
            print(f"\nRunning: BS={batch_size}, DO={dropout}, OPT={opt_name}, LR={lr}, l1_lambda = {l1}, l2_lambda = {l2} model={model_name}")
            
            
            # build model 
            model = build_model(model_name, device, dropout)
            
            # dataloader untuk mendefinisikan aturan batching
            train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=4)
            val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=4)
            
            # optimizer
            if opt_name == 'adam':
                optimizer = optim.Adam(model.parameters(), lr=lr)
            else:
                optimizer = optim.RMSprop(model.parameters(), lr=lr)

            criterion = nn.CrossEntropyLoss()
            
            # train and evaluate seluruh hyperparameter  
            epoch_count, train_loss_values, val_loss_values, f1_train_values, f1_val_values, best_f1, epoch_best_f1, val_loss_best_f1, train_loss_best_f1, best_model_state, total_time_seconds = train_and_evaluate(model_name, model, train_loader, val_loader, optimizer, criterion, device, l1, l2)    
            
            print(f"Training and Validation Done in {total_time_seconds} seconds.")
            
            config = { #variabel config akan selalu di-update setiap hyperparameter berubah
                "batch_size": batch_size,
                "dropout": dropout,
                "optimizer": opt_name,
                "lr": lr,
                "l1_lambda": l1,
                "l2_lambda": l2,
                "f1": best_f1,
                "epoch_best_f1": epoch_best_f1,
            }
            
                        
            # simpan .pth model dari setiap kombinasi hyperparameter
            safe_key = config_key.replace(", ", "_").replace("=", "")
            model_path = os.path.join(models_dir, f"{safe_key}.pth")
            torch.save(best_model_state, model_path)   
            
            # simpan hyperparameter + nama model + f1 score (best dan historical) ke dictionary results
            results[config_key] = {
                "config": config,
                "f1_history": f1_val_values,
                "best_f1": best_f1,
                "train_val_duration": total_time_seconds,
                "model_path": model_path
            }

            # best f1 score hapus dari fungsi train_and_Evaluate. cari best f1 score di sini
            if best_f1 > best_result[model_name]["f1"]: # simpan konfigurasi () yang yield model terbaik FROM EACH architecture
                best_result[model_name]["f1"] = best_f1
                best_result[model_name]["config"] = config # simpen config hyperparameter yang disimpan oleh variabel "config" jika f1-score yang dihasilkan oleh config tsb lebih baik daripada current best f1. berbeda dengan best_model_state yang menyimpan weight, bias, buffer (a.k.a "isi otak" model setelah dilatih dan evaluasi), best_config menyimpan konfigurasi hyperparameter (learning rate, optimizer, dll).
                best_result[model_name]["model_state"] = best_model_state # menyimpan weight, bias, buffer (BatchNorm, dll.) setelah model dilatih dan dievaluasi. berbeda dengan best_config yang menyimpan konfigurasi hyperparameter (learning rate, optimizer, dll) model
                best_result[model_name]["epoch_count"] = epoch_count
                best_result[model_name]["train_loss_values"] = train_loss_values
                best_result[model_name]["val_loss_values"] = val_loss_values
                best_result[model_name]["f1_train_values"] = f1_train_values
                best_result[model_name]["f1_val_values"] = f1_val_values
                best_result[model_name]["train_val_duration"] = total_time_seconds
                
            # update json yang berisi kombinasi hyperparameter + nama model + f1 score (historical dan best)
            log_file = os.path.join(logs_dir, f"results_{model_name}.json")
            tmp_file = log_file + ".tmp"
            with open(tmp_file, "w") as f:
                json.dump(results, f, indent=4)
            os.replace(tmp_file, log_file)

    # simpan hyperparameter configuration dan "isi otak" (bias, weight, dll) model --> DONE FOR EACH MODELS
    # torch.save(best_result[model_name], f"best{model_name}.pth")  --> jadinya pake >>> torch.save(best_result[model_name]["model_state"], best_model_path)
    
    print(f"config of best model of {model_name} architecture = {best_result[model_name]['config']}")
    #print(f"model_state of best model of {model_name} architecture = {best_result[model_name]['model_state'].keys()}") --> gak usah, GEDE bgt sumpahh 
    print(f"best f1 score ever reached of best model of {model_name} architecture = {best_result[model_name]['f1']}")
    
    #Simpan BEST MODEL + SUMMARY JSON
    # simpan best model
    best_model_path = os.path.join(models_dir, f"best_{model_name}.pth")
    torch.save(best_result[model_name]["model_state"], best_model_path)

    # simpan JSON berisi summary dari config (hyperparameter), best f1 score, dan model path dari best model 
    summary = {
        "best_config": {  #di config ada f1 yang nyimpen informasi f1 terbaik suatu model. since kita udah punya best_f1 ("best_f1": best_result[model_name]["f1"]), infonya jadi redundant. jadi exclude aja f1 yang di sini
            "batch_size": best_result[model_name]["config"]["batch_size"],
            "dropout": best_result[model_name]["config"]["dropout"],
            "optimizer": best_result[model_name]["config"]["optimizer"],
            "lr": best_result[model_name]["config"]["lr"],
            "l1_lambda": best_result[model_name]["config"]["l1_lambda"],
            "l2_lambda": best_result[model_name]["config"]["l2_lambda"],
        },
        "best_f1": best_result[model_name]["f1"],
        "train_val_duration": best_result[model_name]["train_val_duration"],
        
        #"model_state_keys": list(best_result[model_name]["model_state"].keys()) --> gak usah, model_state tuh udah dismpen di best_model.pth 
    }

    summary_path = os.path.join(logs_dir, f"best_{model_name}_summary.json")
    with open(summary_path, "w") as f:
        json.dump(summary, f, indent=4)
    
```
